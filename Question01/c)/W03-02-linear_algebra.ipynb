Given the affine model:
\[y = w_1 x_1 + w_2 x_2 + w_3\]
1. Determine the values of the weights $w = (w_1, w_2, w_3)$ that make the model fit all of the points:

- $x_1 =1, x_2= 0, y = 2$
- $x_1 = 0, x_2 = 1, y = 3$
- $x_1 = 1, x_2 = 1, y = 6$

2. Rewrite the affine model as an inner-product using inputs and labels.

```python
#Write your answer here
#t1 w = (3,4,-1)

#t2 [3,4,-1]*[1,0,1] = 2, [3,4,-1]*[0,1,1]=3, [3,4,-1]*[1,1,1]=6
# --- input:
# are the three vectors we can form from the function:
# x_1 = [1,0,1], x_2 = [0,1,1], x_3 = [1,1,1]
# Labels are the corresponding y values from the data
# y = [2, 3, 6]
# y_i = w^T*x_i

# t3 
# Aw = y
# A is the matrix made from the three input vetors:
# A = [
# [1, 0, 1],
# [0, 1, 1],
# [1, 1, 1]
# ] 
# *
# [3, 4, -1]
# =
# [2, 3, 6]


import numpy as np
A = np.array([
    [1, 0, 1],
    [0, 1, 1],
    [1, 1, 1]
])

y = np.array([2, 3, 6])

w = np.linalg.solve(A, y)

print(w)
```

**Task 5 (medium): Determine weights 2📽️**
Rewrite the affine model as an inner-product seperating weights and inputs. If not possible, explain why. 
A:

B:

---

To rewrite these models as an inner product, we look for the form  (or ), where weights and features are separated into two vectors. Specifically, to include the bias term  (or ) within the inner product itself, we use a "bias trick" by appending a  to the feature vector.

Here are the models rewritten in vector notation:

---

### **A**: 
$$y = w_1 \cos(x_1) + w_2 x_2^2 + w_3$$
$$\mathbf{w} = \begin{bmatrix} w_1 \\ w_2 \\ w_3 \end{bmatrix}, \quad \mathbf{\phi(x)} = \begin{bmatrix} \cos(x_1) \\ x_2^2 \\ 1 \end{bmatrix}$$



**Result:** 
$$y = \mathbf{w}^T \mathbf{\phi(x)}$$

### **B:**
$$y = e^{x_1} w_1 + w_2 x_2^2 + w_3$$
$$\mathbf{w} = \begin{bmatrix} w_1 \\ w_2 \\ w_3 \end{bmatrix}, \quad \mathbf{\phi(x)} = \begin{bmatrix} e^{x_1} \\ x_2^2 \\ 1 \end{bmatrix}$$

Result: 
$$y = \mathbf{w}^T \mathbf{\phi(x)}$$

### **C:**

$$y = w_3 + \sin(x_1) w_1 + w_2 \ln(x_2)$$
$$\mathbf{w} = \begin{bmatrix} w_1 \\ w_2 \\ w_3 \end{bmatrix}, \quad \mathbf{\phi(x)} = \begin{bmatrix} \sin(x_1) \\ \ln(x_2) \\ 1 \end{bmatrix}$$
Result: 
$$y = \mathbf{w}^T \mathbf{\phi(x)}$$

### **D:**

$$y = w_1 x_1 x_2 + w_2 x_2 + w_3$$
$$\mathbf{w} = \begin{bmatrix} w_1 \\ w_2 \\ w_3 \end{bmatrix}, \quad \mathbf{\phi(x)} = \begin{bmatrix} x_1 x_2 \\ x_2 \\ 1 \end{bmatrix}$$
Result: 
$$y = \mathbf{w}^T \mathbf{\phi(x)}$$

### **E:**


$$y = w_1 x_1 + w_2 \sin(x_2) + w_3 x_3 + w_4 \cos(x_4) + w_5$$
$$\mathbf{w} = \begin{bmatrix} w_1 \\ w_2 \\ w_3 \\ w_4 \\ w_5 \end{bmatrix}, \quad \mathbf{\phi(x)} = \begin{bmatrix} x_1 \\ \sin(x_2) \\ x_3 \\ \cos(x_4) \\ 1 \end{bmatrix}$$
Result: 
$$y = \mathbf{w}^T \mathbf{\phi(x)}$$

---

### **F:**

**Not strictly possible** as a standard weight-feature inner product because the weight  is shared between a feature(($\sin(x_2)$)) and the bias term.

While you can write it as a product, you cannot separate the weights into a vector of unique parameters  where each element acts independently on a unique feature. If forced into a vector format:

$$y =  w_1 x_1 + w_4\sin(x_2)  + w_2 x_3 + w_3 \cos(x_4) + w_4$$
$$\mathbf{w} = \begin{bmatrix} w_1 \\ w_4 \\ w_2 \\ w_3 \\ w_4 \end{bmatrix}, \quad \mathbf{\phi(x)} = \begin{bmatrix} x_1 \\ \sin(x_2) \\ x_3 \\ \cos(x_4) \\ 1 \end{bmatrix}$$


However, this is usually considered a **constrained** model rather than a standard affine model because the feature  and the constant  are tied to the same coefficient.

# Write your reflections here...

### 1
Rewriting these models into inner products is the first step toward Linear Regression, which is the process of finding the best weights ($\mathbf{w}$) to fit a set of data points.

### 2
We can only make one model. In a real world senario it might be possible to find multiple solutions, then we would need to approximate the best possible soulution. This is a made up example where there is only one answer per equation, in the real world we might have many input data points, where it is not possible to find an exact soulution. Here it is given that the model is affine.

### 3
This model has 3 unknowns ($w_1, w_2, w_3$). When you have exactly 3 unique data points, you have a "square" system of equations (square design matrix) that typically has one unique solution (as we found above). If you add a 4th point (or more), the system becomes overdetermined. 
Two things can happen:
The point is on the line: If the new points happen to lie exactly on the plane $y = 2x_1 + 3x_2 + 1$, the solution remains the same.
The new point is not in the line (Real-world Scenario): In reality, data usually contains "noise." A 4th point might be $(1, 2, 9.1)$ instead of exactly $9.0$. In this case, there is no single set of weights that will make the plane pass through every single point simultaneously.

### 4
We would not be able to make linear regression that fits perfectly on the line, we would have to use least squares to find a regression line where the sum of the squared distance from the points to the line is minimized.

### 5
Linearity means that we can write it up as an inner product. Linearity means that is has to be linear in weights but not in the input vector. We can see it in D where the input vector contains $x_1*x_2$, this is okay as long as the model is linear in the weights. Therefor F is unsolveable because we use the same weight twice.
