**Task 1 (easy): Differentiate the following univariate functions♾️**
1. For $f\left( x \right) = {\left( {6{x^2} + 7x} \right)^4}$ show that $\frac{\partial f}{\partial x} = {{4\left( {12x + 7} \right){{\left( {6{x^2} + 7x} \right)}^3}}}$
2. For $g\left( t \right) = {\left( {4{t^2} - 3t + 2} \right)^{ - 2}}$ show that $\frac{\partial g}{\partial t} = {{ - 2\left( {8t - 3} \right){{\left( {4{t^2} - 3t + 2} \right)}^{ - 3}}}}$
3. For $g_2\left( x \right) = 2\sin \left( {3x + \tan \left( x \right)} \right)$ show that $ \frac{\partial g_2}{\partial x}  = 2\left( {3 + {{\sec }^2}\left( x \right)} \right)\cos \left( {3x + \tan \left( x \right)} \right)$
4. For $g_3\left( x \right) = {{\bf{e}}^{1 - \cos \left( x \right)}}$ show that $\frac{\partial g_3}{\partial x} = \sin (x) \bf{e}^{1 - \cos (x)}$


## Answer:
### 1
$$ f(x) = (6x^2 + 7x)^4 $$
$$ g(x) = 6x^2 + 7x $$
$$ \frac{\partial g}{\partial x} = 12x + 7 $$
$$ \frac{\partial f}{\partial x} = 4(g(x))^3 · (\frac{\partial g}{\partial x})= 4(6x^2+7x)^3 · (12x+7) $$

### 2
$$ g(x) = (4t^2 - 3t +2)^{-2} $$
$$ f(x) = 4t^2 - 3t +2 $$
$$ \frac{\partial f}{\partial x} = 8t-3 $$
$$ \frac{\partial g}{\partial x} = -2(f(x))^{-3} · (\frac{\partial f}{\partial x})= -2(4t^2-3t+2)^{-3}(8t-3) $$

### 3
$$ g_2 (x) = 2sin(3x+tan(x)) $$
$$ f_2 (x) = 3x+tan(x) $$
$$ \frac{\partial f_2}{\partial x} = 3 + sec^2(x) $$
$$ \frac{\partial g_2}{\partial x} = 2cos(f(x)) · (\frac{\partial f_2}{\partial x})= 2cos(3x+tan(x))(3 + sec^2(x))$$

### 4
$$ g_3 (x) = e^{1-cos(x)} $$
$$ f_3 (x) = 1-cos(x) $$
$$ \frac{\partial f_3}{\partial x} = sin(x) $$
$$ \frac{\partial g_3}{\partial x} = e^{f_3(x)} · (\frac{\partial f_3}{\partial x})= e^{1-cos(x)}sin(x)$$

**Task 2 (easy): Find the partial derivatives of the multivariate functions♾️**
1. For $f\left( {x,y,z} \right) = 4{x^3}{y^2} - {{\bf{e}}^z}{y^4} + \frac{{{z^3}}}{{{x^2}}} + 4y - {x^{16}}$ show that  $\frac{\partial f}{\partial x}  = 12x^2 y^2 - \frac{2z^3}{x^3} - 16x^{15}$
2. For $w(x,y,z) = \cos \left( {{x^2} + 2y} \right) - {{\bf{e}}^{4x - {z^{\,4}}y}} + {y^3}$ show that  $\frac{{\partial w}}{{\partial y}} =  -2 \sin(x^2 + 2y) + z^4 e^{4x - z^4 y} + 3y^2$
3. For $ r\left( {x,y} \right) = \frac{{{x^2}}}{{{y^2} + 1}} - \frac{{{y^2}}}{{{x^2} + y}}$ show that  $\frac{{\partial r}}{{\partial x}} = \frac{2x}{y^2 + 1} + \frac{2x y^2}{(x^2 + y)^2}$

## Answer:
### 1
$$ f({x,y,z}) = 4{x^3}{y^2} - {{\bf{e}}^z}{y^4} + \frac{{{z^3}}}{{{x^2}}} + 4y - {x^{16}} $$
$$ \frac{\partial}{\partial x}(4x^3y^2) = 12x^2y^2 $$
$$ \frac{\partial}{\partial x}(e^zy^4) = 0$$
$$ \frac{\partial}{\partial x}(\frac{z^3}{x^2}) = z^3·(-2)x^{-3} = -\frac{2z^3}{x^3} $$
$$ \frac{\partial}{\partial x}(x^{16}) = 16x^{15} $$
$$ \frac{\partial f}{\partial x} = 12x^2y^2 - \frac{2z^3}{x^3}  - 16x^{15}  $$

### 2
$$ w({x,y,z}) = cos(x^2+2y) - e^{4x-z^4y}+y^3 $$
$$ \frac{\partial}{\partial y}(cos(x^2+2y)) = -2sin(x^2+2y) $$
$$ \frac{\partial}{\partial y}(e^{4x-z^4y}) = -z^4e^{4x-z^4y} $$
$$ \frac{\partial}{\partial y}(y^3) = 3y^2 $$
$$ \frac{\partial w}{\partial y} = -2sin(x^2+2y) + z^4e^{4x-z^4y} + 3y^2  $$

### 3
$$ r\left( {x,y} \right) = \frac{{{x^2}}}{{{y^2} + 1}} - \frac{{{y^2}}}{{{x^2} + y}} $$
$$ \frac{\partial}{\partial x}(\frac{x^2}{y^2+1}) = \frac{2x}{y^2+1} $$
$$ \frac{\partial}{\partial y}(\frac{y^2}{x^2+y}) = -\frac{2xy^2}{(x^2+y)^2} $$
$$ \frac{\partial f}{\partial x} = \frac{2x}{y^2+1} +  \frac{2xy^2}{(x^2+y)^2} $$

**Task 4 (medium): Gradient♾️**

- $f_1\left( {x,y} \right) = \frac{x^2}{cos(3x)}  - \frac{{{x^2}}}{{{y^3}}}$
- $f_2\left( {x,y,z} \right) = (x\cos \left( {xy} \right) + {z^2}{y^4} - 7xz)^2$

1. Draw the computational graph (following the chain rule) of the functions.
2. Determine the gradients, $\nabla f_1$ and $\nabla f_2$.

## Answer:
### 2
$$ \nabla f_1 =  \begin{bmatrix} \frac{\partial f_1}{\partial x} \\ \frac{\partial f_1}{\partial y} \end{bmatrix} $$
$$ \frac{\partial f_1}{\partial x} = \frac{2x}{\cos(3x)} + \frac{3x^2\tan(3x)}{\cos(3x)} - \frac{2x}{y^3} $$
$$ \frac{\partial f_1}{\partial y} = \frac{3x^2}{y^4} $$
$$ \nabla f_1 =  \begin{bmatrix} \frac{2x}{\cos(3x)} + \frac{3x^2\tan(3x)}{\cos(3x)} - \frac{2x}{y^3} \\  \frac{3x^2}{y^4} \end{bmatrix} $$

$$ \nabla f_2 =  \begin{bmatrix} \frac{\partial f_2}{\partial x} \\ \frac{\partial f_2}{\partial y} \\ \frac{\partial f_2}{\partial z} \end{bmatrix} $$
$$ \frac{\partial f_2}{\partial x} = 2(x\cos(xy) + z^2y^4 - 7xz)(\cos(xy) - xy\sin(xy) - 7z) $$

$$ \frac{\partial f_2}{\partial z} = 2(x\cos(xy) + z^2y^4 - 7xz)(2zy^4 - 7x) $$
